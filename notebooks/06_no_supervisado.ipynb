```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
import sys
import os

# Importamos las funciones que ya hizo tu compañera
sys.path.append(os.path.abspath(".."))
from src.data_preprocessing import load_data, categorize_quality, prepare_data

# Cargamos los datos exactamente igual
df_raw = load_data("../data/winequality-white.csv")
df_bin = categorize_quality(df_raw)
X_train, X_test, y_train, y_test, scaler = prepare_data(df_bin)

print("Datos cargados. Tamaño de X_train:", X_train.shape)
```

```python
# Instanciamos K-Means pidiendo 3 grupos (porque sabemos que hay Bajo, Medio y Alto)
# n_init=10 significa que el algoritmo intentará 10 veces empezar desde cero para buscar el mejor agrupamiento
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

# fit_predict entrena el modelo y al mismo tiempo nos devuelve a qué grupo asignó cada vino
clusters_predichos = kmeans.fit_predict(X_train)

print("Primeros 20 vinos asignados a los clusters:", clusters_predichos[:20])
```

```python
# Calculamos el puntaje de la silueta
score = silhouette_score(X_train, clusters_predichos)

print(f"Silhouette Score: {score:.4f}")
```
