Implementación de modelos supervisados (clasificación/regresión) con Scikit-learn.

```python
import sys
import os

# Esto permite que el notebook busque archivos en la carpeta raíz 'src'
sys.path.append(os.path.abspath(".."))

from src.data_preprocessing import load_data, categorize_quality, prepare_data
```

```python
# Cargar datos (Sin tocar el original)
df_raw = load_data("../data/winequality-white.csv")

# Categorizar (Para que los modelos aprendan mejor)
df_bin = categorize_quality(df_raw)

# Preparar variables para el entrenamiento
# Esto entrega los datos ya escalados (Media 0, Varianza 1)
X_train, X_test, y_train, y_test, scaler = prepare_data(df_bin)

# Verificación rápida
print(f"Forma de X_train: {X_train.shape}")
print(f"Forma de X_test: {X_test.shape}")
```

```python
# Verificamos cómo quedaron distribuidas las 3 categorías
# 0: Bajo, 1: Medio, 2: Alto
print("Distribución de clases en el dataset procesado:")
print(df_bin['quality'].value_counts().sort_index())

print("\nPorcentaje por categoría:")
print(df_bin['quality'].value_counts(normalize=True).sort_index() * 100)
```

```python
from src.model_training import get_models, train_model

# Obtener los modelos configurados
modelos_dict = get_models(random_state=42)

# Diccionario para guardar los modelos ya entrenados
modelos_entrenados = {}

# Bucle de entrenamiento
for nombre, modelo in modelos_dict.items():
    print(f"Entrenando {nombre}...")
    modelos_entrenados[nombre] = train_model(modelo, X_train, y_train)
    print(f"{nombre} entrenado con éxito.")
```

```python
import joblib
import os

# Creamos la carpeta de destino si no existe
path = '../models/trained_models/'
os.makedirs(path, exist_ok=True)

# Guardamos los modelos para usarlos en el notebook 03
joblib.dump(modelos_entrenados['Random_Forest'], f'{path}rf_model.pkl')
joblib.dump(modelos_entrenados['SVM'], f'{path}svm_model.pkl')
joblib.dump(scaler, f'{path}scaler.pkl')

print("Fase de modelado completada. Modelos exportados a /models/trained_models/")
```
